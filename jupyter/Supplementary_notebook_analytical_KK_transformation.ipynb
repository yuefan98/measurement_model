# Analytical Kramers-Kronig transformation for Randles circuit (RC)

```python
# Define the symbols for the RC circuit parameters
R_ct, w, w_p, eps = sp.symbols('R_ct w w_p, eps')
```

```python
## Define the Real part of RC circuit
def RC_Re (R_ct, w_p):
    return R_ct / (1 + w_p**2)
def RC_Im (R_ct, w_p):
    return -R_ct * w_p / (1 + w_p**2)
```

```python
print('Real part of RC circuit:')
RC_Re (R_ct, w_p)
```

```python
print('Imaginary part of RC circuit:')
RC_Im (R_ct, w_p)
```

## Equation 6: Imaginary to Real Part of the RC Circuit

```python
## Define the KK transformation for RC from Imaginary to Real part
KK_Im_to_Re_RC = (2/sp.pi) * w_p * RC_Im(R_ct,w_p)/ (w**2 - w_p**2)
```

```python
## Find the KK integral
sp.simplify(sp.integrate(KK_Im_to_Re_RC,w_p))
```

```python
## Perform the Cauchy Principal Value integration from 0 to w-eps and w+eps to infinity
int_1 = sp.integrate(KK_Im_to_Re_RC, (w_p, 0, w-eps))
int_2 = sp.integrate(KK_Im_to_Re_RC, (w_p, w+eps, sp.oo))
```

```python
# simplify the lower integral
sp.simplify(int_1)
```

```python
# simplify the upper integral
sp.simplify(int_2)
```

```python
# Combine the two integrals and take the limit as eps approaches 0
# simplify the final result
int = sp.simplify(sp.limit(int_1+int_2, eps, 0))
int
```

```python
# Further simplify the result by replacing log(-w) with log(w) + I*pi 
# to obtain the final expression that matches the expected form
int_simp = sp.simplify(int.xreplace({sp.log(-w): sp.log(w) + sp.I*sp.pi}))
int_simp
```

```python
## Verify the Real part of RC circuit calculated by KK integration 
## matches the exact form of the Real part of the RC circuit
sp.simplify(int_simp-RC_Re(R_ct, w))
```

## Equation 7: Real to Imaginary Part of the RC Circuit

```python
## Define the KK transformation for RC from Real to Imaginary part
KK_Re_to_Im_RC = -(2/sp.pi) * w * RC_Re(R_ct,w_p)/ (w**2 - w_p**2)
```

```python
## Find the KK integral
sp.simplify(sp.integrate(KK_Re_to_Im_RC,w_p))
```

```python
## Perform the Cauchy Principal Value integration from 0 to w-eps and w+eps to infinity
int_1 = sp.integrate(KK_Re_to_Im_RC, (w_p, 0, w-eps))
int_2 = sp.integrate(KK_Re_to_Im_RC, (w_p, w+eps, sp.oo))
```

```python
# simplify the lower integral
sp.simplify(int_1)
```

```python
# simplify the upper integral
sp.simplify(int_2)
```

```python
# Combine the two integrals and take the limit as eps approaches 0
# simplify the final result
int = sp.simplify(sp.limit(int_1+int_2, eps, 0))
int
```

```python
# Further simplify the result by replacing log(-w) with log(w) + I*pi 
# to obtain the final expression that matches the expected form
int_simp = int.xreplace({sp.log(-w): sp.log(w) + sp.I*sp.pi})
int_simp
```

```python
## Verify the Imaginary part of RC circuit calculated by KK integration 
## matches the exact form of the Imaginary part of the RC circuit
sp.simplify(int_simp-RC_Im(R_ct, w))
```

# Analytical Kramers-Kronig transformation for Nonlinear Randles circuit (NLRC)

```python
# Define the symbols for the RC circuit parameters
# R_ct, w, w_p, eps, f, ε = sp.symbols('R_ct w w_p eps f ε')
f, ε = sp.symbols('f ε', real=True)
```

```python
## Define the Real part of RC circuit
def NLRC_Re (R_ct,w_p, ε, f):
    Z1_re = RC_Re(R_ct, w_p)
    Z1_im = RC_Im(R_ct, w_p)
    # return sp.simplify(-(ε * f) * (Z1_re**2 - Z1_im**2 + 4* w_p * Z1_re * Z1_im )/ (1+4*w_p**2))
    Z2_re = -(ε * f) * (Z1_re**2 - Z1_im**2 + 4* w_p * Z1_re * Z1_im )/ (1+4*w_p**2)
    # return sp.apart(sp.simplify(Z2_re), w_p)
    return sp.simplify(Z2_re)


def NLRC_Im (R_ct, w_p, ε, f):
    Z1_re = RC_Re(R_ct, w_p)
    Z1_im = RC_Im(R_ct, w_p)
    return sp.simplify((ε * f) * (2 * w_p * (Z1_re**2 - Z1_im**2) - 2 * Z1_re * Z1_im) / (1+4*w_p**2))
```

```python
print('Real part of nonlinear RC circuit:')
# sp.expand(NLRC_Re(R_ct, w_p, ε, f))
NLRC_Re(R_ct, w_p, ε, f)
```

```python
print('Imaginary part of RC circuit:')
NLRC_Im (R_ct, w_p, ε, f)
```

```python
# the combined form of the nonlinear RC circuit
Z2_combined = NLRC_Re (R_ct, w_p, ε, f) + sp.I * NLRC_Im (R_ct, w_p, ε, f)
```

```python
# the exact form of the nonlinear RC circuit
Z2 = -ε*f * R_ct**2 / (1+4*w_p*sp.I - 5*w_p**2 - 2*w_p**3*sp.I)
```

```python
## Verify the real and imaginary parts of the nonlinear RC circuit can be combined and match the expected form of Z2 for nonlinear RC circuit
sp.simplify(Z2_combined-Z2)
```

## Equation 6: Imaginary to Real Part of the NLRC Circuit

```python
## Define the KK transformation for RC from Imaginary to Real part
KK_Im_to_Re_NLRC = (2/sp.pi) * w_p * NLRC_Im(R_ct,w_p, ε, f)/ (w**2 - w_p**2)
```

```python
# Perform the Cauchy Principal Value integration from 0 to w-eps 
int_1 = sp.integrate(KK_Im_to_Re_NLRC, (w_p, 0, w-eps))
```

```python
# Perform the Cauchy Principal Value integration from w+eps to infinity
int_2 = sp.integrate(KK_Im_to_Re_NLRC, (w_p, w+eps, sp.oo))
```

```python
int = sp.simplify(sp.limit(int_1+int_2, eps, 0, dir='+'))
```

```python
# Further simplify the result by replacing log(-w) with log(w) + I*pi 
# to obtain the final expression that matches the expected form
int_simp = sp.simplify(int.xreplace({sp.log(-w): sp.log(w) + sp.I*sp.pi}))
int_simp
```

```python
## Verify the Real part of NLRC circuit calculated by KK integration 
## matches the exact form of the Real part of the NLRC circuit
sp.simplify(int_simp-NLRC_Re(R_ct, w, ε, f))
```

## Equation 7: Real to Imaginary Part of the RC Circuit

```python
## Define the KK transformation for RC from Real to Imaginary part
KK_Re_to_Im_NLRC = -(2/sp.pi) * w * NLRC_Re(R_ct,w_p,ε,f)/ (w**2 - w_p**2)
```
